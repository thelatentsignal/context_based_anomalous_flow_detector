```python
import torch


def create_even_tensor(n: int) -> torch.tensor:
    # YOUR CODE HERE

    even_tensor = torch.arange(0, 2 * n, 2, dtype=torch.int64, device=torch.device("cpu"))
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(device)
    even_tensor = even_tensor.to(device)
    return even_tensor


# Test your code:
# expected output for n=4: tensor([0., 2., 4., 6.], device=...)
print(create_even_tensor(4))
```
